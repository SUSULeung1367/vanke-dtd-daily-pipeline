# Vanke Daily DTD Date Range Demo

Select a start date and an end date. The notebook processes every calendar day in the inclusive range and prints the Daily Result, Vanke Temporary Input and Temporary DTD tables.

```python
from pathlib import Path
import sys

import pandas as pd

PROJECT_DIR = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(PROJECT_DIR / 'src'))

from vanke_dtd.workflow import run_date_range

START_DATE = "20251213"
END_DATE = "20251219"
MODE = "REPLAY"       # REPLAY is stable; LIVE calls external APIs.
RESET_DEMO = True       # True starts from the confirmed 12 December anchor.
COMPANY = 5338
```

## Run the selected range

One selected day gives one Daily Result row. Closed days remain visible but do not create Temporary Input or DTD.

```python
result = run_date_range(
    START_DATE,
    END_DATE,
    project_dir=PROJECT_DIR,
    mode=MODE,
    reset=RESET_DEMO,
    company=COMPANY,
)

print(f"Processed {len(result.daily_results)} calendar day(s): {result.start_date} to {result.end_date}")
print(f"Mode: {result.mode}")
print(f"Demo workspace: {result.workspace}")

print("\nDAILY RESULT AND QC")
display(result.daily_results)

print("\nVANKE TEMPORARY INPUT")
display(result.temporary_input)

print("\nTEMPORARY DTD")
display(result.temporary_dtd)

print("\nUPDATED CLEAN DTD INPUT (CONFIRMED + TEMPORARY)")
display(result.updated_clean_input.tail(15))
```
