# Demo 1 — Run the Daily DTD Pipeline

Select a start date and an end date. The notebook processes every calendar day in the inclusive range and prints the Daily Result, Vanke Temporary Input and Temporary DTD tables.

```python
from pathlib import Path
import sys

import pandas as pd

PROJECT_DIR = Path.cwd().resolve()
if not (PROJECT_DIR / 'src').is_dir():
    PROJECT_DIR = PROJECT_DIR.parent
sys.path.insert(0, str(PROJECT_DIR / 'src'))

from vanke_dtd_pipeline.daily_pipeline_runner import run_date_range

START_DATE = "20251213"
END_DATE = "20251219"
MODE = "BASIC_TEST"  # BASIC_TEST is fixed and offline; LIVE calls external APIs.
RESET_WORKSPACE = True  # True starts from the confirmed 12 December anchor.
COMPANY = 5338
```

## Run the selected range

One selected day gives one Daily Result row. Closed days remain visible but do not create Temporary Input or DTD.

```python
result = run_date_range(
    START_DATE,
    END_DATE,
    project_dir=PROJECT_DIR,
    workspace_dir=PROJECT_DIR / "runtime" / "basic_test_workspace",
    mode=MODE,
    reset=RESET_WORKSPACE,
    company=COMPANY,
)

print(f"Processed {len(result.daily_results)} calendar day(s): {result.start_date} to {result.end_date}")
print(f"Mode: {result.mode}")
print(f"Basic-test workspace: {result.workspace}")

print("\nDAILY RESULT AND QC")
display(result.daily_results)

print("\nPENDING-REVIEW DTD INPUT")
display(result.pending_review_input)

print("\nPENDING-REVIEW DTD OUTPUT")
display(result.pending_review_dtd_output)

print("\nUPDATED DTD INPUT (CONFIRMED + PENDING REVIEW)")
display(result.updated_clean_input.tail(15))
```
